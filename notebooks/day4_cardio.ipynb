Time, distance, pace, and an estimated-power trajectory, built from what's logged in the app. Cardio logs a single time (min) + distance (mi) entry per session — no sets/reps/weight, so this doesn't use the strength-day volume/power estimates.

**Power (watts) is not measured** — it's a physiological estimate from the ACSM level-ground walking/running metabolic equations (VO2 as a linear function of speed, walking coefficient below ~5 mph and running coefficient above), converted to calories via ~5 kcal/L O2, using the most recent `/weight` entry as body mass. Treat it as a relative trend across sessions, not a lab-accurate number. Tune the constants in `analysis_lib.py` (`add_cardio_power`) if it feels off.
